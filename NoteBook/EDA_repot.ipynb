## Data Cleaning Required
The following columns are stored as strings and should be converted to numeric:

```python
df["growth rate"] = df["growth rate"].str.replace("%","").astype(float)
df["world percentage"] = df["world percentage"].str.replace("%","").astype(float)
```

1. Top 10 Most Populated Countries (2023)

```python
top_10_countries=df.sort_values(by="2023 population",ascending=False).head(10)
sns.barplot(x="2023 population",y="country",data=top_10_countries)
plt.title("Top 10 Countries by population in 2023")
plt.xlabel("Population")
plt.ylabel("Country")
plt.show()
```

2. Population Distribution

```python
sns.histplot(df["2023 population"],bins=10,kde=True)
plt.title("Distribution of 2023 Population")
plt.xlabel("Population")
plt.ylabel("Frequency")
plt.show
```

Replaced a column name

```python
df.rename(columns={"density (km²)":"density (km2)"},inplace=True)
```

```python
df.columns
```

3. Population vs Density

```python
sns.scatterplot(x="2023 population",y="density (km2)",data=df)
plt.title("Population and Density Scatter plot")
plt.xlabel("2023 Population")
plt.ylabel("Density")
```

4. Population by Continent

```python
continent_pop = df.groupby("continent")["2023 population"].sum().reset_index()
sns.barplot(x="continent",y="2023 population",data=continent_pop)
plt.xticks(rotation=45)
plt.show()
```

5. Growth Rate Analysis

```python
top_growth = df.nlargest(10, "growth rate")
sns.barplot(x="growth rate",y="country",data=top_growth)
plt.title("Highest Growth Rate Countries")
plt.show()
```

6. Correlation Heatmap

```python
numeric_df = df.select_dtypes(include=["int64","float64"])
plt.figure(figsize=(12,8))
sns.heatmap(numeric_df.corr(),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

7. Population Growth (1970 → 2023)

```python
df["population_growth"] = (df["2023 population"]-df["1970 population"])
df.nlargest(10,"population_growth")[["country","population_growth"]]
```

8. Density Distribution

```python
sns.boxplot(x=df["density (km2)"])
plt.title("Density Outliers")
plt.show()
```

## Key Insights
1.The world population is highly concentrated within a few countries.

2.Asia contributes the largest share of the global population.

3.Historical population figures strongly predict current population levels.

4.Population density varies dramatically across nations due to differences in land area and urbanization.

5.Many developing countries continue to experience higher growth rates than developed nations.
